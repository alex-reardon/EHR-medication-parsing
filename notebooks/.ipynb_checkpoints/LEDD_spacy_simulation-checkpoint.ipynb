```python
import pandas as pd
import re 
import matplotlib.pyplot as plt
from rapidfuzz import process, fuzz
import numpy as np
import unicodedata 
import os
import spacy 
```

```python
dir = '/Users/emudr/Desktop/'
outdir = '/Users/emudr/Desktop/'
file = 'PPMI_LEDTRT_StructuredStyle_20_Subjects.csv'
df = pd.read_csv(dir+file)

```

```python
nlp = spacy.load("en_core_web_sm")  # SciSpaCy model


def add_sentence_column(df, source_col, new_col="sentence_split"):
    def split_text(text):
        if pd.isna(text):
            return []
        doc = nlp(str(text))
        return [sent.text.strip() for sent in doc.sents]
    
    df[new_col] = df[source_col].apply(split_text)
    return df

# Apply
df = add_sentence_column(df, "LEDTRT", "LEDTRT_sentences")
# Save safely
outfile = os.path.join(outdir, "sentence_sg.csv")
df.to_csv(outfile, index=False)
```
