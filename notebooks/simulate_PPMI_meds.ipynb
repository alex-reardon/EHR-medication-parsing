```python
import pandas as pd
import re 
import numpy as np
import unicodedata 
import random
```

```python
dir = '/Users/emudr/PPMI_LEDD/data/PPMI/'
raw_file = 'raw/LEDD_Concomitant_Medication_Log.csv'
df = pd.read_csv(dir + raw_file, low_memory = False)
```

```python
daily_phrases = [
    "x per day",
    "times per day",
    "x a day",
    "a day", 
    "daily",
    "each day",
    "times every day",
    "times everyday",
    "x every day", 
    "per day",
    "per 24 hours",
    "every 24 hours"
]

form_phrases = [
    'pill', 
    'pills', 
    'capsule', 
    'capsules',
    'tab', 
    'tabs', 
    'softgel',
    'dose']


def add_phrase(df, col,  phrase_list):
    def _add(freq):
        if pd.isna(freq):
            return None
        return f"{float(freq)} {random.choice(phrase_list)}"

    df[col] = df[col].apply(_add)
    return df


df = add_phrase(df, 'LEDDOSFRQ', daily_phrases)
df = add_phrase(df, 'LEDDOSE', form_phrases) 
```

```python
def combine_row(row):
    return " ".join(
        str(x) for x in row
        if pd.notna(x) and str(x).strip() != ""
    )

cols = ["LEDTRT", "LEDDOSSTR", "LEDDOSE", "LEDDOSFRQ"]
df["LEDTRT_simulated"] = df[cols].apply(combine_row, axis=1)

col_reorder = ['REC_ID',	'PATNO','EVENT_ID', 'PAG_NAME',	'INFODT', 'SEQNO', 'LEDTRT_simulated', 'LEDTRT', 'LEDDSTRMG', 'LEDDOSSTR', 'LEDDOSE', 'LEDDOSFRQ', 'STARTDT', 'STOPDT',	'LEDD',	'ORIG_ENTRY', 'LAST_UPDATE']
df = df[col_reorder]
df.to_csv(dir + 'simulated/LEDD_Concomitant_Medication_Log_simulated.csv')
```
